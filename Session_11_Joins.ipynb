- In Pandas we can join two dataframes

    - merge

    - join

    - concat

**merge**

- it is under pandas the function: pd.merge

- it is similar to SQL joins

- you can join with specific columns

```python
import pandas as pd
```

```python
d1={'ID':[1,2,3],'Name':['Chandu','Pawan','Bansal']}
d2={'ID':[1,2,4],'Age':[20,25,30]}
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on='ID',how='inner')
```

```python
pd.merge(df1,df2,on='ID',how='outer')
```

```python
pd.merge(df1,df2,on='ID',how='left')
```

```python
pd.merge(df1,df2,on='ID',how='right')
```

```python
## Homework

df1.join(df2)

pd.concat([df1,df2])
```
