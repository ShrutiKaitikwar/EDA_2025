- I want seperate Catgorical and numerical columns

```python
# idea
# make two empty list cat_list,num_list
# iterate the data_types.items zip(keys,values)
# if value==int or float then key should go to numlist

cat=[]
num=[]
for key,value in zip(Keys,Values):
    if value=='object':
        cat.append(key)
    else:
        num.append(key)
```

```python
cat=[]
num=[]
for key,value in dict(data_types).items():
    if value=='object':
        cat.append(key)
    else:
        num.append(key)
```
