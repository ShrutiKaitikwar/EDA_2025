- Get outliers data and with out data seperate

```python
q1=np.quantile(wage_data,0.25)
q2=np.quantile(wage_data,0.50)
q3=np.quantile(wage_data,0.75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
con1=wage_data<lb
con2=wage_data>ub
con=con1 | con2
len(visa_df[con])
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
con1=wage_data<lb
con2=wage_data>ub
con=con1 | con2
outliers_df=visa_df[con]
len(visa_df[con])   # with outliers data
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
ub=q3+1.5*iqr
con1=wage_data>lb
con2=wage_data<ub
con=con1 & con2
non_outliers_data=visa_df[con]
len(visa_df[con])   # with out outliers data
```
